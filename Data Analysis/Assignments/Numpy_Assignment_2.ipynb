### 8. Create two arrays of shape (2, 3): arr1 and arr2. Perform element-wise addition, subtraction, multiplication, and division on them.

```python
array1 = np.random.randint(10,100,(2,3))
array2 = np.random.randint(10,100,(2,3))
print(array1)
print(array2)
print("Addition\n",array1 + array2)
print("Subtraction\n",array1 - array2)
print("Multipliication\n",array1 * array2)
print("Division\n",array1 / array2)
```

### 9. Use NumPy's built-in functions to compute the sum, mean, and standard deviation of a given array.

```python
arr9 = np.arange(10,101,10)
print(arr9)
print(arr9.sum())
print(arr9.mean())
print(arr9.std())
```

### 10. Create a 2x2 matrix A and a 2x2 matrix B. Perform matrix multiplication using np.dot() or the @ operator.

```python
A = np.arange(10,14).reshape(2,2)
B = np.arange(20,24).reshape(2,2)
print(A)
print(B)
print(A @ B)
```

### 13. Create a 4x4 array of random integers. Find the maximum, minimum, and mean of the entire array.

### 14. Find the row-wise and column-wise sums.

```python
arr14 = np.arange(10,91,10).reshape(3,3)
print(arr14)
print("Row-Wise Sum :",np.sum(arr14, axis=1))
print("Column-Wise Sum :",np.sum(arr14, axis=0))
```
