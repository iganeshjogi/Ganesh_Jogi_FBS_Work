### Q3. Perform the following operations on an array of mobile phones prices 6999, 7500, 11999, 27899, 14999, 9999.

#### a. Create a 1d-array of mobile phones prices

```python
mo_price = np.array([6999, 7500, 11999, 27899, 14999, 9999])
mo_price
```

#### b. Convert this array to float type

```python
mo_price = mo_price.astype(float)
mo_price
```

#### c. Append a new mobile having price of 13999 Rs. to this array

```python
mo_price = np.append(mo_price, 13999)
mo_price
```

#### d. Reverse this array of mobile phones prices

```python
mo_price = mo_price[::-1]
mo_price
```

#### e. Apply GST of 18% on mobile phones prices and update this array.

```python
mo_price = mo_price + (mo_price * 18/100)
mo_price
```

#### f. Sort the array in descending order of price

```python
mo_price = np.sort(mo_price)[::-1]
mo_price
```

#### g. What is the average mobile phone price

```python
np.average(mo_price)
```

#### h. What is the difference b/w maximum and minimum price

```python
np.max(mo_price)  - np.min(mo_price)
```
