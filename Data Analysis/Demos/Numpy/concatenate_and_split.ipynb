## Concatenation and Split

```python
arr1 = np.arange(1, 21).reshape((5, 4))
arr1
```

```python
arr2 = np.arange(21, 41).reshape((5, 4))
arr2
```

```python
np.concatenate((arr1, arr2))
```

```python
np.concatenate((arr1, arr2),axis=1)
```

## Spilt

```python
np.split(arr1, 2, axis=1)
```
