# Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
import pandas as pd
```

### Bar Plot
1. Comparing the categories
2. showing frequencies or count
3. eg. product sales comparison

```python
drinks = ['hot coffee', 'cold coffee', 'tea', 'salad']
summer = [30, 80, 70, 50]
```

```python
plt.figure(figsize=(7,3))
plt.bar(drinks, summer, color = 'orange', label='summer', linestyle='--', linewidth=1, edgecolor='blue')
plt.title('Sales')
plt.xlabel('Drinks')
plt.ylabel('Summer')
```

#### Histogram
1. Distribution of numerical data
2. Frequency of values ranges
3. eg. skewness, normal distribution, outliners

```python
Male_age = np.random.randint(15, 40, 100)
Female_age = np.random.randint(15, 40, 100)
```

```python
plt.hist([Male_age, Female_age], label=['Male', 'Female'])
plt.title('Age distribution')
plt.xlabel('Age Range')
plt.ylabel('Frequencies')
plt.legend()
plt.show()
```

### Line plot
1. Showing trends over time
2. Time series analysis
3. Upward/Downward trends
4. Seasonal patterns

```python
team1_runs = [0, 10, 15, 18, 38, 45, 60, 68, 79, 85, 100]
team2_runs = [0, 8, 14, 13, 34, 30, 50, 78, 75, 32, 120]
overs = np.arange(0, 11)
```

```python
plt.figure(figsize=(7,3))
plt.plot(overs, team1_runs, label='Team 1', color='orange', marker='o')
plt.plot(overs, team2_runs, label='Team 2', color='blue', marker='<')
plt.xlabel('Overs')
plt.ylabel('Runs')
plt.title('Score card')
plt.legend()
plt.show
```

# Pie plot
    1.Showing percentage distribution
    2. part-to-whole comparison

```python
data = [50, 40, 30, 60]
product = ['Cold Coffee', 'Hot Coffee', 'Salad', 'Tea']
explode = [0.01, 0.02, 0.05, 0.02]
```

```python
plt.pie(data, labels=product, autopct='%0.2f%%', explode=explode)
plt.show()
```
